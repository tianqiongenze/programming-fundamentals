### Starter 2：`def check_device(dev, temp):`
（Investigate: 读代码→Modify: 补全TODO）

```
def check_device(dev, temp):
    """TODO-2 判断级别: ≥95→3 / ≥80→2 / 其他→0(正常)
    返回级别数字(0-3)"""
    # 【参考答案】TODO-2 与w03()内check_level一致: ≥95→3级 / ≥80→2级 / 其他→0
    return 3 if temp >= 95 else (2 if temp >= 80 else 0)
```

### Starter 3：`def alarm(dev, level, temp):`
（Investigate: 读代码→Modify: 补全TODO）

```
def alarm(dev, level, temp):
    """TODO-3 输出告警并累计到stats (格式自定, 要含设备id与温度)"""
    # 【参考答案】TODO-3 输出含设备id/温度的告警行并累计stats(与w03()循环体一致)
    if level:
        stats[dev["id"]][level] += 1
        print(f"[{dev['id']}]{dev['name']} 温度{temp:.1f}℃ → "
              f"{'警告' if level == 2 else '停机'}")
```

### Starter 4：`if __name__ == "__main__":`
（Investigate: 读代码→Modify: 补全TODO）

```
if __name__ == "__main__":
    # 【参考答案】TODO-4 双层循环 3设备×10周期: base_temp + random.uniform(-8, 15)
    for d in devices:
        for _ in range(10):
            t = d["base_temp"] + random.uniform(-8, 15)   # 制造偶发超限
            lv = check_device(d, t)
            alarm(d, lv, t)
    # 【参考答案】TODO-5 循环结束输出《告警统计报告》(与w03()统计口径一致)
    print("== 告警统计 ==")
    for k, v in stats.items():
        print(k, "警告", v[2], "停机", v[3])
```

## 教师参考答案（来源：`教师参考/solutions.py` · `w03()`）

> **发放纪律：本附录勿随学生版下发。** 课堂按层放提示（D层演示照做/C层只给思路行/A层直接做）。

学生版 TODO 位置 → 下方参考实现的对应关系：

| 学生版 TODO 函数 | 参考实现说明 |
|---|---|
| `check_device()` | 见下方 `w03()` 内对应逻辑 |
| `alarm()` | 见下方 `w03()` 内对应逻辑 |

```
def w03():
    random.seed(42)
    devices = [{"id": "DEV-01", "name": "空压机A", "base": 62},
               {"id": "DEV-02", "name": "冷却泵", "base": 55},
               {"id": "DEV-03", "name": "注塑机", "base": 70}]
    stats = {d["id"]: {2: 0, 3: 0} for d in devices}

    def check_level(temp):
        return 3 if temp >= 95 else (2 if temp >= 80 else 0)

    for d in devices:
        for _ in range(10):
            t = d["base"] + random.uniform(-8, 15)
            lv = check_level(t)
            if lv:
                stats[d["id"]][lv] += 1
                print(f"[{d['id']}]{d['name']} 温度{t:.1f}℃ → "
                      f"{'警告' if lv == 2 else '停机'}")
    print("== 告警统计 ==")
    for k, v in stats.items():
        print(k, "警告", v[2], "停机", v[3])
```
