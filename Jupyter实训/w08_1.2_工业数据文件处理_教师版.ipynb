### Starter 3：`def main():`
（Investigate: 读代码→Modify: 补全TODO）

```
def main():
    ok, bad_rows = 0, []
    clean_rows, total = [], 0
    # 【参考答案】TODO-2 with open读取RAW, 逐行validate(与w08()主循环一致)
    with open(RAW, encoding="utf-8") as f:
        for row in csv.DictReader(f):
            total += 1
            err = validate(row)
            if err is None:
                ok += 1
                clean_rows.append(row)
            else:
                bad_rows.append([row["device_id"], row["temp"], err])
    # 【参考答案】TODO-3 坏案例写入BAD(带表头可追溯), 干净行写入CLEAN(与w08()输出口径一致)
    with open(BAD, "w", newline="", encoding="utf-8") as f:
        w = csv.writer(f)
        w.writerow(["device_id", "temp", "err"])
        for r in bad_rows:
            w.writerow(r)
    with open(CLEAN, "w", newline="", encoding="utf-8") as f:
        w = csv.writer(f)
        w.writerow(["device_id", "temp", "press", "status"])
        for r in clean_rows:
            w.writerow([r["device_id"], r["temp"], r["press"], r["status"]])
    # 【参考答案】TODO-4 清洗报告: 总数/成功/各错误类型计数(与w08()报告口径一致)
    counter = {"缺失": 0, "类型": 0, "范围": 0}
    for _, _, e in bad_rows:
        counter[e] += 1
    print(f"清洗报告: 总{total} 成功{ok} 问题{len(bad_rows)} {counter}")
```

## 教师参考答案（来源：`教师参考/solutions.py` · `w08()`）

> **发放纪律：本附录勿随学生版下发。** 课堂按层放提示（D层演示照做/C层只给思路行/A层直接做）。

学生版 TODO 位置 → 下方参考实现的对应关系：

| 学生版 TODO 函数 | 参考实现说明 |
|---|---|
| `validate()` | 见下方 `w08()` 内对应逻辑 |
| `main()` | 见下方 `w08()` 内对应逻辑 |

```
def w08():
    import csv
    import os
    random.seed(7)
    raw = "raw_demo.csv"
    if not os.path.exists(raw):
        with open(raw, "w", newline="", encoding="utf-8") as f:
            w = csv.writer(f)
            w.writerow(["device_id", "temp", "press", "status"])
            for i in range(60):
                t = round(random.uniform(40, 90), 1)
                if i % 17 == 3:
                    t = ""
                if i % 23 == 5:
                    t = "err"
                if i % 29 == 7:
                    t = 135.0
                w.writerow([f"DEV-{i % 3 + 1:02d}", t,
                            round(random.uniform(.3, .6), 2), "RUN"])

    def validate(row):
        t = row["temp"]
        if t in ("", None):
            return "缺失"
        try:
            t = float(t)
        except ValueError:
            return "类型"
        if not 0 <= t <= 120:
            return "范围"
        return None

    ok, bad, counter = 0, [], {"缺失": 0, "类型": 0, "范围": 0}
    with open(raw, encoding="utf-8") as f:
        for row in csv.DictReader(f):
            err = validate(row)
            if err is None:
                ok += 1
            else:
                bad.append((row, err))
                counter[err] += 1
    with open("bad_records.csv", "w", newline="", encoding="utf-8") as f:
        w = csv.writer(f)
        w.writerow(["device_id", "temp", "err"])
        for row, e in bad:
            w.writerow([row["device_id"], row["temp"], e])
    print(f"清洗报告: 总60 成功{ok} 问题{len(bad)} {counter}")
```
