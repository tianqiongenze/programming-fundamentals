### Starter 1：`# -*- coding: utf-8 -*-`

```python
# -*- coding: utf-8 -*-
"""W11 工业互联网数据采集系统 —— Modbus采集 + MQTT上报
前置: 先启动虚拟产线(见README): mini_modbus_server.py 5020 1 与 mini_mqtt_broker.py
"""
import json
import time

from pymodbus.client import ModbusTcpClient as Client
import paho.mqtt.client as mqtt

PLC_HOST, PLC_PORT = "127.0.0.1", 5020
BROKER = "127.0.0.1"
TOPIC = "device/1/status"
NAMES = ["temp_c", "press_mpa", "vib_mm_s", "rpm", "status"]
```

### Starter 2：`def read_plc():`
（Investigate: 读代码→Modify: 补全TODO）

```python
def read_plc():
    """TODO-1 完成态：连接PLC读5个寄存器, ÷10还原工程值(status除外)(与w11()一致)"""
    # 【参考答案】TODO-1 连接PLC读5个寄存器, ÷10还原工程值(status除外)(与w11()一致)
    c = Client(host=PLC_HOST, port=PLC_PORT)
    try:
        if not c.connect():
            return None
        rr = c.read_holding_registers(address=0, count=5)
        vals = rr.registers
        return {n: (v / 10 if n != "status" else v)
                for n, v in zip(NAMES, vals)}
    except Exception:
        return None
    finally:
        c.close()
```

### Starter 3：`def publish(mq, data):`
（Investigate: 读代码→Modify: 补全TODO）

```python
def publish(mq, data):
    """TODO-2 完成态：JSON序列化+ts质量戳+发布到TOPIC(与w11()payload结构一致)"""
    # 【参考答案】TODO-2 JSON序列化+ts质量戳+发布到TOPIC(与w11()payload结构一致)
    payload = json.dumps({"device_id": "DEV-1",
                          "ts": time.strftime("%H:%M:%S"),
                          "metrics": data}, ensure_ascii=False)
    mq.publish(TOPIC, payload)
    print("上报:", payload)
```

### Starter 4：`def main():`
（Investigate: 读代码→Modify: 补全TODO）

```python
def main():
    # paho-mqtt 2.x要求显式指定callback API版本（VERSION1保持经典on_connect签名）
    mq = mqtt.Client(mqtt.CallbackAPIVersion.VERSION1)
    # 【参考答案】TODO-3 连接Broker(connect/loop_start), 失败时try捕获(与w11()一致)
    try:
        mq.connect(BROKER, 1883)
        mq.loop_start()
    except Exception as exc:
        print("MQTT连接失败:", exc)
        return
    for i in range(5):
        data = read_plc()
        if data is None:
            print("采集失败, 重试...")
            time.sleep(1)
            continue
        print("第%02d轮:" % (i + 1), data)
        publish(mq, data)
        time.sleep(2)   # 【参考答案】TODO-4(决策) 间隔取2秒: 兼顾采集频率与broker负载(写入工单)
    # 【参考答案】TODO-5(优秀) 订阅TOPIC回环验证: on_message打印回执
    # 注: executed版(w11())未包含订阅回环, 此处按工单TODO描述补写
    def on_message(client, userdata, msg):
        print("回环回执:", msg.topic, msg.payload.decode("utf-8", "ignore"))
    mq.on_message = on_message
    mq.subscribe(TOPIC)
    time.sleep(1)       # 等待回环回执到达
    mq.loop_stop()
```
