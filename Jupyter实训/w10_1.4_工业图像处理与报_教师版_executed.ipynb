### Starter 1：`# -*- coding: utf-8 -*-`

```python
# -*- coding: utf-8 -*-
"""W10 工业图像处理与报告生成 —— Pillow标注 + reportlab PDF"""
import os

from PIL import Image, ImageDraw
from reportlab.lib.pagesizes import A4
from reportlab.lib.units import mm
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.platypus import (Image as RLImage, Paragraph, SimpleDocTemplate,
                                Spacer, Table, TableStyle)
from reportlab.lib.styles import getSampleStyleSheet

import os as _os
FONT = next((f"fonts/{n}" for n in ("ZH_Font.ttf", "SimHei.ttf")
           if _os.path.exists(f"fonts/{n}")), "SimHei.ttf")
pdfmetrics.registerFont(TTFont("ZH", FONT))          # 中文字体注册(勿删)
pdfmetrics.registerFont(TTFont("SimHei", FONT))

# 缺陷信息: (x, y, w, h, 说明)
DEFECTS = [(120, 80, 90, 40, "划痕 9.0mm"), (220, 150, 45, 45, "凹坑 4.5mm²")]
```

### Starter 2：`def make_marked_image(src, dst):`
（Investigate: 读代码→Modify: 补全TODO）

```python
def make_marked_image(src, dst):
    """TODO-1 完成态：缩放统一尺寸+红框与文字标注(标注数据来自DEFECTS, 与w10()画法一致)"""
    # 【参考答案】TODO-1 缩放统一尺寸+红框与文字标注(标注数据来自DEFECTS, 与w10()画法一致)
    img = Image.open(src).resize((400, 300))
    d = ImageDraw.Draw(img)
    for x, y, w, h, note in DEFECTS:
        d.rectangle([x, y, x + w, y + h], outline=(255, 0, 0), width=3)
        d.text((x, y - 14), note, fill=(255, 0, 0))
    img.save(dst)
    return dst
```

### Starter 3：`def build_pdf(img_path):`
（Investigate: 读代码→Modify: 补全TODO）

```python
def build_pdf(img_path):
    """TODO-2 完成态：a~2d 标题+日期/结论段/标注图/GRID统计表(与w10()PDF组装一致)"""
    doc = SimpleDocTemplate("质量日报.pdf", pagesize=A4,
                            leftMargin=20 * mm, rightMargin=20 * mm)
    styles = getSampleStyleSheet()
    styles["Title"].fontName = "SimHei"
    # 【参考答案】TODO-2a~2d 标题+日期/结论段/标注图/GRID统计表(与w10()PDF组装一致)
    styles["Normal"].fontName = "SimHei"          # 正文样式同样用中文字体
    import datetime
    story = [
        # 【参考答案】TODO-2a 标题: 《设备质量日报》+日期(日期为工单TODO要求, executed版无日期)
        Paragraph("设备质量日报 " + datetime.date.today().isoformat(),
                  styles["Title"]),
        # 【参考答案】TODO-2b 结论段落(2-3句)
        Paragraph("结论: 本批次共发现缺陷2处(划痕1处/凹坑1处), "
                  "其中划痕9.0mm超出判定标准, 建议复检; "
                  "凹坑4.5mm²在允收范围内。", styles["Normal"]),
        # 【参考答案】TODO-2c 标注图
        RLImage(img_path, width=120 * mm, height=90 * mm),
        # 【参考答案】TODO-2d 统计表: 缺陷/尺寸/判定 三列, TableStyle含GRID
        Table([["缺陷", "尺寸", "判定"], ["划痕", "9.0mm", "超差"],
               ["凹坑", "4.5mm²", "合格"]],
              style=TableStyle([("GRID", (0, 0), (-1, -1), 0.5, "black"),
                                ("FONTNAME", (0, 0), (-1, -1), "SimHei")])),
    ]
    doc.build(story)
    return "质量日报.pdf"
```

### Starter 4：`if __name__ == "__main__":`
（Investigate: 读代码→Modify: 补全TODO）

```python
if __name__ == "__main__":
    img = make_marked_image("sample_device.png", "marked.png") \
        if os.path.exists("sample_device.png") else None
    build_pdf(img or "marked.png")
    print("已生成 质量日报.pdf")
```

## 教师参考答案（来源：`教师参考/solutions.py` · `w10()`）

> **发放纪律：本附录勿随学生版下发。** 课堂按层放提示（D层演示照做/C层只给思路行/A层直接做）。

学生版 TODO 位置 → 下方参考实现的对应关系：

| 学生版 TODO 函数 | 参考实现说明 |
|---|---|
| `make_marked_image()` | 见下方 `w10()` 内对应逻辑 |
| `build_pdf()` | 见下方 `w10()` 内对应逻辑 |

```python
def w10():
    from PIL import Image, ImageDraw
    from reportlab.lib.pagesizes import A4
    from reportlab.lib.units import mm
    from reportlab.pdfbase import pdfmetrics
    from reportlab.pdfbase.ttfonts import TTFont
    from reportlab.platypus import (Image as RLImage, Paragraph,
                                    SimpleDocTemplate, Table, TableStyle)
    from reportlab.lib.styles import getSampleStyleSheet

    pdfmetrics.registerFont(TTFont("SimHei", "fonts/ZH_Font.ttf"))
    img = Image.new("RGB", (400, 300), (200, 200, 200))
    d = ImageDraw.Draw(img)
    d.rectangle([120, 80, 210, 120], outline=(255, 0, 0), width=3)
    d.text((120, 66), "scratch 9.0mm", fill=(255, 0, 0))
    d.rectangle([220, 150, 265, 195], outline=(255, 0, 0), width=3)
    d.text((220, 136), "pit 4.5mm2", fill=(255, 0, 0))
    img.save("marked.png")

    st = getSampleStyleSheet()
    st["Title"].fontName = st["Normal"].fontName = "SimHei"
    tbl = Table([["缺陷", "尺寸", "判定"], ["划痕", "9.0mm", "超差"],
                 ["凹坑", "4.5mm²", "合格"]])
    tbl.setStyle(TableStyle([("GRID", (0, 0), (-1, -1), 0.5, "black"),
                             ("FONTNAME", (0, 0), (-1, -1), "SimHei")]))
    SimpleDocTemplate("质量日报.pdf", pagesize=A4).build([
        Paragraph("设备质量日报", st["Title"]), Paragraph(
            "结论: 发现划痕1处超差, 建议复检。", st["Normal"]),
        RLImage("marked.png", width=120 * mm, height=90 * mm), tbl])
    print("已生成 质量日报.pdf")
```
