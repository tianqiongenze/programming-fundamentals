```python
# -*- coding: utf-8 -*-
"""W09 工业故障分析报告生成系统 —— jieba分词+词云"""
import os
from collections import Counter

import jieba
import jieba.analyse
import matplotlib.pyplot as plt

plt.rcParams["font.sans-serif"] = ["Arial Unicode MS", "SimHei", "Heiti TC"]  # 中文字体

TEXTS = [
    "1号注塑机液压油温升高报警，影响生产节拍", "3号车床主轴轴承磨损异响，班中巡检发现",
    "空压机B排气压力不足，请尽快处理", "冷却水泵2机械密封泄漏，已挂牌停机",
    "包装机A的PLC通信中断，从站掉线", "5号压铸机液压油污染度超标，滤芯堵塞",
    "辊道输送线电机电流偏高，轴承座发热", "中央冷水机组冷却水温持续偏高，填料结垢",
    "变频器过流报警，负载卡滞需检查", "行车2电缆绝缘老化，接地故障",
    "AGV充电站温度变送器读数漂移，偏差超限", "数控车床冷却系统流量不足，温差增大",
    "注塑机液压电磁阀卡滞，换向失灵", "空压机振动剧烈伴随杂音，轴承磨损",
    "包装线气动调节阀振荡，定位器故障", "压铸机液压油缸内泄，保压不住",
] * 3   # ×3扩成48条(可再合并班级语料)

STOP = set("的 了 在 是 请 尽快 处理 影响 生产 班中 巡检 发现 已 挂牌 停机 需 检查".split())
import os as _os
FONT = next((f"fonts/{n}" for n in ("ZH_Font.ttf", "SimHei.ttf")
           if _os.path.exists(f"fonts/{n}")), "SimHei.ttf")  # 资料包字体(含CJK)
```

### Starter 2：`def main():`
（Investigate: 读代码→Modify: 补全TODO）

```python
def main():
    # 【参考答案】TODO-1 分词+停用词过滤 → words列表(与w09()一致, 语料用本工单TEXTS/STOP)
    words = [w for t in TEXTS for w in jieba.cut(t)
             if w not in STOP and len(w) > 1]
    # 【参考答案】TODO-2 词频统计(Counter) + Top10柱状图(保存bar.png; 与w09()一致)
    top10 = Counter(words).most_common(10)
    print("Top10:", top10)
    labels = [w for w, _ in top10][::-1]
    counts = [c for _, c in top10][::-1]
    fig, ax = plt.subplots(figsize=(7, 4.2), dpi=120)
    bars = ax.barh(labels, counts, color="#2b6cb0")
    ax.set_title("工业故障描述 Top10 关键词词频")
    ax.set_xlabel("词频")
    ax.set_xlim(0, max(counts) * 1.2)
    for bar, cnt in zip(bars, counts):
        ax.text(bar.get_width() + max(counts) * 0.02,
                bar.get_y() + bar.get_height() / 2, str(cnt),
                va="center", fontsize=9)
    fig.tight_layout()
    fig.savefig("bar.png")
    plt.close(fig)
    print("已保存 bar.png")
    # 【参考答案】TODO-3 jieba.analyse.extract_tags关键词(topK按工单取20, 含权重)
    for kw, wt in jieba.analyse.extract_tags(" ".join(TEXTS), topK=20,
                                             withWeight=True)[:8]:
        print(f"关键词 {kw:<6} 权重{wt:.4f}")
    # 【参考答案】TODO-4 wordcloud生成词云(保存cloud.png; 与w09()一致含零依赖兜底)
    try:
        from wordcloud import WordCloud
        wc = WordCloud(font_path=FONT, width=800, height=400,
                       background_color="white").generate(" ".join(words))
        wc.to_file("cloud.png")
        print("词云实现: wordcloud(font_path=%s)" % FONT)
    except Exception as exc:
        print("[w09] wordcloud 不可用 → matplotlib 标签云零依赖兜底:", exc)
        from matplotlib import cm
        fig, ax = plt.subplots(figsize=(9, 4.5), dpi=120)
        ax.set_axis_off()
        ys = [0.85, 0.63, 0.41, 0.19]
        for i, (word, cnt) in enumerate(top10):
            ax.text(0.05 + (i % 5) * 0.19, ys[i // 5] + 0.02,
                    f"{word}×{cnt}", fontsize=10 + 14 * cnt / max(counts),
                    color=cm.tab10(i % 10), transform=ax.transAxes)
        ax.set_title("工业故障关键词标签云(matplotlib零依赖兜底, 字号∝词频)")
        fig.tight_layout()
        fig.savefig("cloud.png")
        plt.close(fig)
    print("已保存 cloud.png")
    # 【参考答案】TODO-5 (洞察)打印Top10词并给出≤150字故障规律分析(供写入工单)
    print("故障规律分析: 高频词集中于 液压/轴承/报警/冷却/注塑机 等, 故障多发于"
          "液压系统与旋转部件(轴承/电机/密封), 以温升、磨损、泄漏类为主; 建议将"
          "点检重点放在液压油质与轴承状态, 提前更换滤芯与密封件, 降低突发停机。")
```

```python
def w09():
    import jieba
    import jieba.analyse
    from collections import Counter
    texts = ["1号注塑机液压油温升高报警", "3号车床主轴轴承磨损异响",
             "空压机排气压力不足", "PLC通信中断从站掉线",
             "冷却水温持续偏高填料结垢", "变频器过流报警负载卡滞"] * 3
    stop = set("的 了 在 是 请 尽快 影响 报警 需".split())
    words = [w for t in texts for w in jieba.cut(t)
             if w not in stop and len(w) > 1]
    print("Top10:", Counter(words).most_common(10))
    print("关键词:", jieba.analyse.extract_tags(" ".join(texts), topK=8))
```
