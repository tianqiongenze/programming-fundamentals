### Starter 2：`class Device:`
（Investigate: 读代码→Modify: 补全TODO）

```
class Device:
    """通用设备基类"""

    def __init__(self, dev_id, name):
        self.dev_id = dev_id
        self.name = name
        self.status = "停止"

    def start(self):
        self.status = "运行"
        print(f"{self.name} 已启动")

    def stop(self):
        self.status = "停止"
        print(f"{self.name} 已停止")

    def report(self):
        return f"{self.dev_id} {self.name}: {self.status}"
```

### Starter 3：`# TODO-1 PLC子类: 新增属性scan_ms(扫描周期), 覆盖start()输出`
（Investigate: 读代码→Modify: 补全TODO）

```
# 【参考答案】TODO-1 PLC子类: 新增属性scan_ms(扫描周期), 覆盖start()输出含扫描周期信息
class PLC(Device):
    """PLC子类: __init__用super()调用父类, 再加scan_ms"""

    def __init__(self, dev_id, name, scan_ms=10):
        """super()调父类, 再加scan_ms扫描周期"""
        super().__init__(dev_id, name)
        self.scan_ms = scan_ms

    def start(self):
        """覆盖start: 先调super(), 再输出扫描周期信息"""
        super().start()                       # super()协作(与w05()一致)
        print(f"  扫描周期: {self.scan_ms}ms")
```

### Starter 4：`# TODO-2 变频器子类: 新增方法set_freq(f)(限制0-50Hz, 超范围提`
（Investigate: 读代码→Modify: 补全TODO）

```
# 【参考答案】TODO-2 变频器子类: 新增方法set_freq(f)(限制0-50Hz, 超范围提示)
class VFD(Device):
    """变频器子类: 新增频率设定方法"""

    def set_freq(self, f):
        """设定频率0-50Hz, 超范围提示(与w05()一致)"""
        if 0 <= f <= 50:
            self.freq = f
            print(f"{self.name} 频率设定 {f}Hz")
        else:
            print("频率超范围(0-50Hz)")
```

## 教师参考答案（来源：`教师参考/solutions.py` · `w05()`）

> **发放纪律：本附录勿随学生版下发。** 课堂按层放提示（D层演示照做/C层只给思路行/A层直接做）。

学生版 TODO 位置 → 下方参考实现的对应关系：

| 学生版 TODO 函数 | 参考实现说明 |
|---|---|

```
def w05():
    class Device:
        def __init__(self, dev_id, name):
            self.dev_id, self.name, self.status = dev_id, name, "停止"

        def start(self):
            self.status = "运行"
            print(f"{self.name} 已启动")

        def report(self):
            return f"{self.dev_id} {self.name}: {self.status}"

    class PLC(Device):
        def __init__(self, dev_id, name, scan_ms=10):
            super().__init__(dev_id, name)
            self.scan_ms = scan_ms

        def start(self):                      # 覆盖
            super().start()
            print(f"  扫描周期: {self.scan_ms}ms")

    class VFD(Device):
        def set_freq(self, f):
            if 0 <= f <= 50:
                self.freq = f
                print(f"{self.name} 频率设定 {f}Hz")
            else:
                print("频率超范围(0-50Hz)")

    devices = [PLC("DEV-01", "S7-1200", 8), VFD("DEV-02", "ABB变频器"),
               Device("DEV-03", "温度变送器")]
    for d in devices:                         # 多态: 统一调用不同行为
        d.start()
        print(d.report())
    devices[1].set_freq(35)
```
