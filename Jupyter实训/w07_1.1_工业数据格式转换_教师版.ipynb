```
# -*- coding: utf-8 -*-
"""W07 工业数据格式转换工具 —— 容器+正则(六步法: 决策→实施)"""
import json
import re

# 三种原始格式样例(可扩充)
MSG = "DEV-01:temp=56.3C;press=0.45MPa;status=RUN"
CSV_LINE = "DEV-02,55.1,0.44,RUN"
JSON_STR = '{"device_id": "DEV-03", "temp": 54.8, "press": 0.46, "status": "RUN"}'
```

### Starter 2：`def parse_message(msg):`
（Investigate: 读代码→Modify: 补全TODO）

```
def parse_message(msg):
    r"""TODO-1 正则解析报文串 → 标准字典
    目标: {"device_id":"DEV-01","metrics":{"temp":56.3,"press":0.45,"status":"RUN"}}
    提示: re.match(r'([A-Z]+-\d+):', msg)取ID; re.findall(r'(\w+)=([\d.]+|\w+)', ...)取键值
    数值自动转float(用try或判断)"""
    # 【参考答案】TODO-1 re.match取ID + re.findall取键值 + 数值转float(与w07()一致)
    m = re.match(r"([A-Z]+-\d+):", msg)
    dev = m.group(1) if m else None
    metrics = {}
    for k, v in re.findall(r"(\w+)=([\d.]+|[A-Z]+)", msg):
        metrics[k] = float(v) if re.fullmatch(r"[\d.]+", v) else v
    return {"device_id": dev, "metrics": metrics}
```

### Starter 3：`def parse_csv(line):`

```
def parse_csv(line):
    """TODO-2 CSV行解析 → 同样的标准字典"""
    # 【参考答案】TODO-2 四列拆分+数值转float(与w07()一致)
    dev, t, p, s = line.split(",")
    return {"device_id": dev,
            "metrics": {"temp": float(t), "press": float(p), "status": s}}
```

### Starter 4：`def parse_json(text):`
（Investigate: 读代码→Modify: 补全TODO）

```
def parse_json(text):
    """TODO-3 json.loads后重组为标准字典(metrics嵌套)"""
    # 【参考答案】TODO-3 json.loads后pop device_id重组(与w07()一致)
    data = json.loads(text)
    return {"device_id": data.pop("device_id"), "metrics": data}
```
