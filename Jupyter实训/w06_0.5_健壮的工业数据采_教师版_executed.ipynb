### Starter 2：`def read_lines(path):`
（Investigate: 读代码→Modify: 补全TODO）

```python
def read_lines(path):
    """TODO-2: 打开文件, FileNotFoundError→ERROR日志+返回[];
    逐行yield字符串"""
    # 【参考答案】TODO-2 with open逐行yield; 文件缺失→ERROR日志+空生成器(见w06()读取逻辑)
    try:
        with open(path, encoding="utf-8") as f:
            for line in f:
                yield line.strip("\n")
    except FileNotFoundError:
        logger.error("文件不存在: %s", path)
        return []
```

### Starter 3：`def parse(line):`
（Investigate: 读代码→Modify: 补全TODO）

```python
def parse(line):
    """TODO-3: 拆分temp,press并转float; 值错误→ValueError由上层捕获"""
    # 【参考答案】TODO-3 拆分两列并转float; 脏数据抛ValueError交上层捕获(与w06()一致)
    t, p = line.split(",")
    return float(t), float(p)
```

### Starter 4：`def main():`
（Investigate: 读代码→Modify: 补全TODO）

```python
def main():
    ok = skip = fail = 0
    bad_rows = []
    for line in read_lines(RAW):
        if line.startswith("temp"):
            continue
        # 【参考答案】TODO-4 try/except/except/finally 完整结构(与w06()口径一致)
        try:
            parse(line)
            logger.info("采集成功: %s", line)
            ok += 1
        except ValueError:
            logger.warning("脏数据跳过: %r", line)
            skip += 1
            bad_rows.append(line)
        except Exception as exc:
            logger.error("采集异常: %r (%s)", line, exc)
            fail += 1
        finally:
            pass
    print(f"采集统计: 成功{ok} 跳过{skip} 失败{fail}")
    # 【参考答案】TODO-5(优秀) 脏数据写入bad_records.csv(坏案例留档思路同w08())
    with open("bad_records.csv", "w", encoding="utf-8") as f:
        f.write("raw_line\n")
        for ln in bad_rows:
            f.write(f"{ln}\n")
    print(f"脏数据已留档: bad_records.csv 共{len(bad_rows)}行")
```
